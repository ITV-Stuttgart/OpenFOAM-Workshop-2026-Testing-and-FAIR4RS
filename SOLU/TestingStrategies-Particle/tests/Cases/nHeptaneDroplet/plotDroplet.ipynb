```python
# Plot the droplet evaporation 

import matplotlib.pyplot as plt
import numpy as np

```

```python
# ==============================================================================
# n-Heptane Droplet Evaporation
# ==============================================================================


# Load data
data_Sim = np.genfromtxt('dropletDiameter_nHeptane.dat')
expData = np.genfromtxt('expNHeptaneDroplet.dat',delimiter=',',skip_header=2)

# Initial droplet diameter
d0=7E-4


fSize = 22
lw = 2.5
plt.rc('text', usetex=True)
plt.rc('text.latex',preamble=r'\usepackage{amsmath}')
plt.rc('font', family='serif')
plt.rc('font', size=fSize)
plt.rc('figure',figsize=[7,6])



plt.figure()
plt.plot(data_Sim[:,0]/np.power(d0*1000,2),np.power(data_Sim[:,1]/d0,2),label='Sim.')
plt.plot(expData[:,0],expData[:,1],ls='None',marker='o',label='Exp. (Nomura et al.)')
plt.xlabel(r'$(t/d_0^2) [s/mm^2]$')
plt.ylabel(r'$(d/d_0)^2$')
plt.legend(fontsize='xx-small')
plt.grid(color='lightgray')
plt.xlim([0,7.5])
plt.ylim([0,1])
plt.tight_layout()
plt.savefig('nHeptaneDropletEvap.png',dpi=300,format='png')  

plt.figure()
plt.plot(data_Sim[:,0]/np.power(d0*1000,2),data_Sim[:,2],label='Sim.')
plt.ylim([290,380])
plt.xlabel(r'$(t/d_0^2) [s/mm^2]$')
plt.ylabel(r'$T [K]$')
plt.legend(fontsize='xx-small')
plt.grid(color='lightgray')
plt.xlim([0,7.5])
plt.tight_layout()
plt.savefig('nHeptaneDropletEvap-T.png',dpi=300,format='png')  
```
